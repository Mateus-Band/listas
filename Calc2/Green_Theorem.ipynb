## Teorema de Green
- Define que o quanto que uma região é afetada por um campo, no seu interior, é igual a o quanto é afetada em sua borda.
- Formula (Sendo D a região e C sua borda):
$$\oint_C (P \, dx + Q \, dy) = \iint_D \left( \frac{\partial Q}{\partial x} - \frac{\partial P}{\partial y} \right) \, dA$$


```python
#Código para visualização:
#1. Calcula ambos os lados da equação para um exemplo específico, mostrando que são iguais.
#2. Gera uma visualização da região, da curva fronteira e do campo vetorial.

var('x y t')

P(x, y) = -y
Q(x, y) = x^3 + x*y

xt(t) = cos(t)
yt(t) = sin(t)

t0 = 0
tf = 2*pi

dx = diff(xt(t), t)
dy = diff(yt(t), t)

integrando_linha = P(xt(t), yt(t)) * dx + Q(xt(t), yt(t)) * dy

lado_esquerdo = integral(integrando_linha, t, t0, tf)

print(f'Campo vetorial F = ({P(x,y)}, {Q(x,y)})')
print(f"Curva C: r(t) = ({xt(t)}, {yt(t)}) para t em [{t0}, {tf}]")

print(f'\nResultado do lado esquerdo = {lado_esquerdo}')

#lado direito agora
rotacional = diff(Q, x) - diff(P, y)

print(f"\nLado direito \nIntegrando (dQ/dx - dP/dy) = {rotacional(x,y)}")

lado_direito = integral(integral(rotacional, y, -sqrt(1-x^2), sqrt(1-x^2)), x, -1, 1)

print(f'\nResultado do lado direito = {lado_direito}')




#1. Plot da Região D
grafico_regiao = region_plot(x^2 + y^2 <= 1, (x, -1.5, 1.5), (y, -1.5, 1.5),
                             incol='lightgreen', bordercol='red', alpha=0.3)

#2. Plot da borda C
grafico_borda = parametric_plot((xt(t), yt(t)), (t, t0, tf),
                                color='blue', thickness=2, linestyle='-')

#adicionando uma seta para mostrar a orientação anti-horária (orientação positiva)
seta_orientacao = arrow((xt(pi/4), yt(pi/4)), (xt(pi/4 + 0.1), yt(pi/4 + 0.1)), color='blue')

#3. Plot do campo F
grafico_campo = plot_vector_field((P(x,y), Q(x,y)), (x, -1.2, 1.2), (y, -1.2, 1.2),
                                  color='gray')

grafico_final = grafico_regiao + grafico_campo + grafico_borda + seta_orientacao
grafico_final.show(aspect_ratio=1, title="Visualização para o Teorema de Green")
```
